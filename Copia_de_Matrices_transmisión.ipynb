Reducción de kron

```python
import numpy as np

def kron_reduction(Z):
    """
    Realiza la reducción de Kron eliminando los conductores de guarda (últimas dos filas y columnas).
    :param Z: Matriz de impedancias original (NxN).
    :return: Matriz reducida sin los conductores de guarda.
    """
    # Índices de los nodos a eliminar (últimos dos en este caso)
    n = Z.shape[0]
    idx_remove = [n-2, n-1]  # Últimos dos índices para las guardas

    # Particionar la matriz en bloques
    Zaa = np.delete(np.delete(Z, idx_remove, axis=0), idx_remove, axis=1)  # Parte que nos interesa
    Zab = np.delete(Z[:, idx_remove], idx_remove, axis=0)  # Interacción entre conductores y guardas
    Zbb = Z[np.ix_(idx_remove, idx_remove)]  # Impedancia de las guardas

    # Invertir Zbb
    Zbb_inv = np.linalg.inv(Zbb)

    # Aplicar reducción de Kron: Z_reducida = Zaa - Zab @ Zbb_inv @ Zab.T
    Z_reducida = Zaa - Zab @ Zbb_inv @ Zab.T

    return Z_reducida

# Definir la matriz de impedancias complejas (8x8) basada en tu tabla
Z = np.array([
    [1.4199 + 1j*890.23,  0.0592 + 1j*400.66,  0.0592 + 1j*348.41,  0.0592 + 1j*309.24,  0.0592 + 1j*331.86,  0.0592 + 1j*336.84,  0.7362 + 1j*0.3554,  0.0592 + 1j*0.3153],
    [0.0592 + 1j*400.66,  1.4199 + 1j*890.23,  0.0592 + 1j*400.66,  0.0592 + 1j*331.86,  0.0592 + 1j*336.84,  0.0592 + 1j*331.86,  0.0592 + 1j*0.32254, 0.0592 + 1j*0.3005],
    [0.0592 + 1j*348.41,  0.0592 + 1j*400.66,  1.4199 + 1j*890.23,  0.0592 + 1j*336.84,  0.0592 + 1j*331.86,  0.0592 + 1j*309.24,  0.0592 + 1j*0.2998,  0.0592 + 1j*0.2844],
    [0.0592 + 1j*309.24,  0.0592 + 1j*331.86,  0.0592 + 1j*336.84,  1.4199 + 1j*890.23,  0.0592 + 1j*400.66,  0.0592 + 1j*348.41,  0.0592 + 1j*0.2844,  0.0592 + 1j*0.2998],
    [0.0592 + 1j*331.86,  0.0592 + 1j*336.84,  0.0592 + 1j*331.86,  0.0592 + 1j*400.66,  1.4199 + 1j*890.23,  0.0592 + 1j*400.66,  0.0592 + 1j*0.3005, 0.0592 + 1j*0.32554],
    [0.0592 + 1j*336.84,  0.0592 + 1j*331.86,  0.0592 + 1j*309.24,  0.0592 + 1j*348.41,  0.0592 + 1j*400.66,  1.4199 + 1j*890.23,  0.0592 + 1j*0.3153,  0.0592 + 1j*0.554],
    [0.7362 + 1j*0.3554,  0.0592 + 1j*0.32254, 0.0592 + 1j*0.2998,  0.0592 + 1j*0.2844,  0.0592 + 1j*0.3005, 0.0592 + 1j*0.3153,  0.7362 + 1j*890.2268,  0.0592 + 1j*0.3306],
    [0.0592 + 1j*0.3153,  0.0592 + 1j*0.3005,  0.0592 + 1j*0.2844,  0.0592 + 1j*0.2998,  0.0592 + 1j*0.32254,  0.0592 + 1j*0.3554,  0.0592 + 1j*0.3306,  0.7362 + 1j*0.2268]
])

# Aplicar reducción de Kron
Z_reducida = kron_reduction(Z)

# Imprimir la nueva matriz reducida
print("Matriz de impedancias reducida:")
print(Z_reducida)
```

Matriz lambda

```python
import numpy as np

# Definir la matriz original
lambda_matrix = np.array([
    [8.119, 1.819, 0.8243, 0.3291, 0.8655, 0.8243, 2.8822, 1.0209],
    [1.819, 8.099, 1.231, 0.5334, 0.5334, 0.3185, 9.422, 7.076],
    [0.8243, 1.231, 8.119, 0.2941, 1.231, 0.3291, 4.913, 3.738],
    [0.3291, 0.5334, 0.2941, 8.119, 0.5334, 0.8243, 3.738, 4.913],
    [0.8655, 0.5334, 1.231, 0.5334, 8.099, 1.231, 7.076, 9.422],
    [0.8243, 0.3185, 0.3291, 0.8243, 1.231, 8.119, 1.0209, 2.8822],
    [2.8822, 9.422, 4.913, 3.738, 7.076, 1.0209, 8.119, 8.8002],
    [1.0209, 7.076, 3.738, 4.913, 9.422, 2.8822, 8.8002, 8.119]
])

# Filas y columnas a eliminar (G1 y G2)
rows_to_remove = [6, 7]
cols_to_remove = [6, 7]

# Extraer la submatriz de nodos que quedan (FF)
lambda_FF = np.delete(lambda_matrix, rows_to_remove, axis=0)
lambda_FF = np.delete(lambda_FF, cols_to_remove, axis=1)

# Extraer la submatriz de interacciones (FG)
lambda_FG = lambda_matrix[np.ix_([0, 1, 2, 3, 4, 5], [6, 7])]

# Extraer la submatriz de nodos eliminados (GG)
lambda_GG = lambda_matrix[6:, 6:]

# Calcular la inversa de la matriz GG
lambda_GG_inv = np.linalg.inv(lambda_GG)

# Aplicar la reducción de Kron
lambda_KF = lambda_FF - lambda_FG @ lambda_GG_inv @ lambda_FG.T

# Mostrar resultados
print("Matriz Original:")
print(lambda_matrix)
print("\nMatriz KFF:")
print(lambda_KF)
```

Desbalance matriz de impedancias

```python
import numpy as np

# Datos de la matriz de impedancias extraídos del taller (sin cables de guarda)
Z_matrix = np.array([
    [1.4199 + 1j*890.23, 0.0592 + 1j*400.66, 0.0592 + 1j*348.41],
    [0.0592 + 1j*400.66, 1.4199 + 1j*890.23, 0.0592 + 1j*331.86],
    [0.0592 + 1j*348.41, 0.0592 + 1j*331.86, 1.4199 + 1j*890.23]
])

# Cálculo del desbalance porcentual
Z_ii = np.diag(Z_matrix)  # Elementos propios (diagonal)
Z_ij = np.triu(Z_matrix, 1)  # Elementos mutuos superiores

# Diferencias relativas con la diagonal
desbalance_percent = np.abs(Z_ij - Z_ij.T) / np.abs(Z_ii) * 100

# Obtener el mayor desbalance
max_desbalance = np.nanmax(desbalance_percent)  # nanmax ignora los valores NaN

max_desbalance
```
